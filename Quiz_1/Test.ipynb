```python
# Question:

# You are given a dataset containing the education levels of both parents of each child:

# mother_education → years of schooling completed by the mother

# father_education → years of schooling completed by the father

# For each child, use the highest education level between the mother and the father, and then categorize it into one of the following groups:

# Less than secondary school → fewer than 10 years

# Senior secondary school → exactly 12 years

# More than senior secondary school but not a college graduate → 13 to 15 years

# College graduate or more → 16 years or more

# Part A
# Write a function proportion_of_parental_education that calculates the fraction of children in each category.
# The function should return a dictionary in this format (values will depend on dataset):

# {
#     "less than secondary school": 0.2,
#     "Senior secondary school": 0.4,
#     "more than senior secondary school but not college graduate": 0.2,
#     "College graduate or more": 0.2
# }


# Part B
# Using the dictionary returned from Part A, create a bar chart that shows the proportions of children in each education category.

# X-axis: Education categories

# Y-axis: Proportion (fraction between 0 and 1)

# Add a title and axis labels

# Rotate X-axis labels slightly for readability
```

```python
def proportion_of_parental_education(data):
    
    less_than_secondary = 0
    senior_secondary = 0 
    more_than_secondary = 0
    college_graduate = 0
    
    for mother_education, father_education in data:
        highest_education = max(mother_education, father_education)

        if highest_education < 10:
            less_than_secondary += 1
        elif highest_education == 12:
            senior_secondary += 1
        elif 13 <= highest_education <= 15:
            more_than_secondary += 1
        else:  # highest_education >= 16
            college_graduate += 1
    
    total_children = len(data)
    
    # Create the result dictionary separately
    result_dict = {
        "less than secondary school": less_than_secondary / total_children,
        "Senior secondary school": senior_secondary / total_children,
        "more than senior secondary school but not college graduate": more_than_secondary / total_children,
        "College graduate or more": college_graduate / total_children
    }
    
    return result_dict


# Part B 
import matplotlib.pyplot as plt

def plot_education_proportions(result_dict):
    
    categories = list(result_dict.keys())
    proportions = list(result_dict.values())
    
    # Create bar chart
    plt.figure(figsize=(10, 6))
    plt.bar(categories, proportions, color='lightblue', edgecolor='black')
    
    # Add title and axis labels
    plt.title('Proportion of Children by Parental Education Category', fontsize=14)
    plt.xlabel('Education Categories', fontsize=12)
    plt.ylabel('Proportion (fraction between 0 and 1)', fontsize=12)
    
    # Rotate X-axis labels for readability
    plt.xticks(rotation=20, ha='right')
    
    # Show the chart
    plt.tight_layout()
    plt.show()
```

```python
# Sample data to test the functions
sample_data = [
    (8, 12),   # max = 12 → Senior secondary school
    (16, 15),  # max = 16 → College graduate or more  
    (9, 7),    # max = 9 → Less than secondary school
    (14, 13),  # max = 14 → More than senior secondary but not college graduate
    (18, 16),  # max = 18 → College graduate or more
    (11, 12),  # max = 12 → Senior secondary school
    (6, 8),    # max = 8 → Less than secondary school
    (15, 14),  # max = 15 → More than senior secondary but not college graduate
    (17, 19),  # max = 19 → College graduate or more
    (10, 12),  # max = 12 → Senior secondary school
    (5, 9),    # max = 9 → Less than secondary school
    (13, 14),  # max = 14 → More than senior secondary but not college graduate
    (20, 18),  # max = 20 → College graduate or more
    (12, 11),  # max = 12 → Senior secondary school
    (7, 6)     # max = 7 → Less than secondary school
]

print("Sample dataset contains", len(sample_data), "children")
print("\nTesting Part A:")

# Test Part A function
result = proportion_of_parental_education(sample_data)

print("Education Category Proportions:")
for category, proportion in result.items():
    print(f"  {category}: {proportion:.3f} ({proportion*100:.1f}%)")

print("\nTesting Part B:")
print("Creating bar chart...")

# Test Part B function  
plot_education_proportions(result)
```
