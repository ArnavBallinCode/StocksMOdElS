Dataset details

https://www.kaggle.com/datasets/bushraqurban/world-education-dataset
The dataset consists of one CSV file named world_education_data.csv with the following columns:

country: The name of the country for which the data is reported
country_code: The ISO country code
year: The year for which the data is reported
gov_exp_pct_gdp: Government expenditure on education as a percentage of GDP
lit_rate_adult_pct: Literacy rate for adults (ages 15 and above) as a percentage of the population
pri_comp_rate_pct: Primary completion rate as a percentage of the relevant age group
pupil_teacher_primary: Pupil-teacher ratio at the primary education level
pupil_teacher_secondary: Pupil-teacher ratio at the secondary education level
school_enrol_primary_pct: Primary school enrollment rate as a percentage of the population

Q1. Using the column gov_exp_pct_gdp:

Find the average government expenditure (as % of GDP) across all countries in 2020.

Identify the top 5 countries with the highest government spending on education.

Plot a bar chart for these top 5 countries.

```python
#pre process the dataset, like just see missing values and stuff  /Users/arnavangarkar/Desktop/Arnav/IIIT/DSP/Excercize_5/world-education-data.csv
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('/Users/arnavangarkar/Desktop/Arnav/IIIT/DSP/Excercize_5/world-education-data.csv')
#chech for total missing values 
df.isnull().sum()
```

```python
#fill the missing values with mean of that column
df.fillna(method ='ffill', inplace=True)
df.isnull().sum()
```

```python
df.fillna(method ='bfill', inplace=True)
df.isnull().sum()
```

```python
# Q1. Using the column gov_exp_pct_gdp:

# Find the average government expenditure (as % of GDP) across all countries in 2020.
avg_gov_exp_2020 = df['gov_exp_pct_gdp'].mean()
print(f"Average government expenditure (as % of GDP) across all countries in 2020: {avg_gov_exp_2020}")

# Identify the top 5 countries with the highest government spending on education.
top_5_countries = df.nlargest(5, 'gov_exp_pct_gdp')[['country', 'gov_exp_pct_gdp']]
print("Top 5 countries with the highest government spending on education:")
print(top_5_countries)

# Plot a bar chart for these top 5 countries.
```
