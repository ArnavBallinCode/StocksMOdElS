```python
#import  /Users/arnavangarkar/Desktop/Arnav/IIIT/DSP/Excercize_4/Online Retail.xlsx
import pandas as pd
import numpy as np

df = pd.read_excel("/Users/arnavangarkar/Desktop/Arnav/IIIT/DSP/Excercize_4/Online Retail.xlsx")

display(df.head())
```

```python
X = df[df['Country'] == 'United Kingdom'].copy()

# Fill missing CustomerIDs
X['CustomerID'] = X['CustomerID'].ffill()

# Clean Description
X['Description'] = (
    X['Description']
    .str.replace(r'[^A-Za-z0-9 ]+', '', regex=True)
    .str.upper()
    .str.strip()
)


#Use str.replace() to clean and standardize text in the Description
# column.
# Replace negative values in the Quantity column with NaN or appropriate
# values.

X.loc[X['Quantity'] < 0, 'Quantity'] = np.nan
```
