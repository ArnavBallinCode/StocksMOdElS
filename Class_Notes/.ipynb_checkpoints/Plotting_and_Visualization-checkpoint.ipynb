linspace and random.randn() and fig with all axes and without axes

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. linspace- a function within the NumPy library used to generate an array of evenly spaced numbers over a specified interval.
x1 = np.linspace(0, 10, 100) # numbers between 0 and 10, spread evenly.
y1 = np.sin(x1)

# 2. Random scatter with randn
x2 = np.random.randn(100) # Generates n random numbers from a standard normal distribution (mean=0, std=1).
y2 = np.random.randn(100) # Generates n random numbers from a standard normal distribution (mean=0, std=1).

# fig, ax = plt.subplots(1, 2, figsize=(10, 4))
# Creates a Figure (fig) and all Axes (ax) at once. Easier for multiple subplots. ax can be single object or array depending on grid.
fig = plt.figure(figsize=(10, 4))

ax1 = fig.add_subplot(1, 2, 1)  # row=1, col=2, position=1
ax2 = fig.add_subplot(1, 2, 2)  # row=1, col=2, position=2

# Left: linspace
# ax[0].plot(x1, y1, label="sin(x)")
ax1.plot(x1, y1, label="sin(x)")
# ax[0].set_title("Using linspace")
ax1.set_title("Using linspace")
# ax[0].legend()
ax1.legend()

# Right: randn
# ax[1].scatter(x2, y2, color='r', alpha=0.6)
ax2.scatter(x2, y2, color='r', alpha=0.6)
# ax[1].set_title("Using randn")
ax2.set_title("Using randn")

plt.show()
```

fig with all axes and either write nrows=2, ncols=3 or just write 2,3. figsize= (width, height)- Python's matplotlib library, the values represent width and height in inches.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

# 2 rows, 3 columns
# fig, ax = plt.subplots(nrows=2, ncols=3, figsize=(10, 6))
fig, ax = plt.subplots(2, 3, figsize=(10, 6))

ax[0, 0].plot(x, np.sin(x))
ax[1, 2].plot(x, np.cos(x))

plt.show()
```

By default x and y values

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.sin(x) * 2
y3 = np.cos(x)
y4 = np.cos(x) * 2

# 2 rows × 2 cols, share x and y across all subplots
fig, ax = plt.subplots(2, 2, figsize=(8, 6))

ax[0, 0].plot(x, y1, label="sin(x)")
ax[0, 0].legend()

ax[0, 1].plot(x, y2, 'r', label="2*sin(x)")
ax[0, 1].legend()

ax[1, 0].plot(x, y3, 'g', label="cos(x)")
ax[1, 0].legend()

ax[1, 1].plot(x, y4, 'm', label="2*cos(x)")
ax[1, 1].legend()

# Common title
fig.suptitle("By default: x and y values", fontsize=14)

plt.show()
```

all subplots use same x and y values

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.sin(x) * 2
y3 = np.cos(x)
y4 = np.cos(x) * 2

# 2 rows × 2 cols, share x and y across all subplots
fig, ax = plt.subplots(2, 2, sharex=True, sharey=True, figsize=(8, 6))

ax[0, 0].plot(x, y1, label="sin(x)")
ax[0, 0].legend()

ax[0, 1].plot(x, y2, 'r', label="2*sin(x)")
ax[0, 1].legend()

ax[1, 0].plot(x, y3, 'g', label="cos(x)")
ax[1, 0].legend()

ax[1, 1].plot(x, y4, 'm', label="2*cos(x)")
ax[1, 1].legend()

# Common title
fig.suptitle("sharex=True, sharey=True", fontsize=14)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.sin(x) * 2
y3 = np.cos(x)
y4 = np.cos(x) * 2

# 2 rows × 2 cols, share x and y across all subplots
fig, ax = plt.subplots(2, 2, sharex=True, sharey=False, figsize=(8, 6))

ax[0, 0].plot(x, y1, label="sin(x)")
ax[0, 0].legend()

ax[0, 1].plot(x, y2, 'r', label="2*sin(x)")
ax[0, 1].legend()

ax[1, 0].plot(x, y3, 'g', label="cos(x)")
ax[1, 0].legend()

ax[1, 1].plot(x, y4, 'm', label="2*cos(x)")
ax[1, 1].legend()

# Common title
fig.suptitle("sharex=True, sharey=False", fontsize=14)

plt.show()
```

Normally, a subplot is a Cartesian plane (x-axis left–right, y-axis up–down).
But sometimes, you want your plot in polar coordinates:

Instead of (x, y), you use (θ, r):

θ = angle (goes around a circle)

r = radius (distance from center)

```python
import matplotlib.pyplot as plt
import numpy as np

theta = np.linspace(0, 2*np.pi, 100) # angles
r = np.abs(np.sin(5*theta)) # radius value

# 2×2 grid, all subplots are polar
fig, ax = plt.subplots(2, 2, subplot_kw={'projection': 'polar'}, figsize=(8, 6)) # Make all subplots polar plots

ax[0, 0].plot(theta, r)
ax[0, 1].scatter(theta, r, c='r')
ax[1, 0].plot(theta, np.cos(5*theta))
ax[1, 1].bar(theta, r)

fig.suptitle("Using subplot_kw={'projection':'polar'}", fontsize=14)
plt.show()
```

subplot_kw={'box_aspect': 1}

This is about the shape of the subplot box.

By default, matplotlib stretches/shrinks plots to fit into the figure grid.

Sometimes, you want your plot to be a perfect square (equal width and height).

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

# All subplots have box_aspect=1 (square)
fig, ax = plt.subplots(1, 2, subplot_kw={'box_aspect': 1}, figsize=(8, 4))

ax[0].plot(x, y)
ax[1].scatter(x, y, c='g')

fig.suptitle("Using subplot_kw={'box_aspect':1}", fontsize=14)
plt.show()
```

Adjusting the spacing around subplots

```python
import matplotlib.pyplot as plt
import numpy as np

# Create a 2x2 grid of subplots with shared X and Y axes
fig, axes = plt.subplots(2, 2, sharex=True, sharey=True)

# Loop through each subplot and plot a histogram
for i in range(2):
    for j in range(2):
        axes[i, j].hist(np.random.randn(500), bins=50, color='k', alpha=0.5)

# Remove spacing between subplots
plt.subplots_adjust(wspace=0.1, hspace=0.2)

plt.show()
```

Ticks, Labels, and Legends

xlim() → Controls the range of x-axis

Without arguments: tells you the current x-axis range.

With arguments: sets the x-axis range.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)

print("Current x-axis range:", plt.xlim())   # returns current range

plt.xlim([0, 15])   # set range manually
plt.show()
```

xticks() → Controls the location of ticks (marks on x-axis)

Without arguments: shows where the ticks are currently.

With arguments: lets you set new tick positions.

```python
plt.plot(x, y)

print("Current tick locations:", plt.xticks())

plt.xticks([0, 2, 4, 6, 8, 10])   # custom ticks
plt.show()
```

```python
plt.plot(x, y)

ticks = [0, 2, 4, 6, 8, 10]
labels = ['zero', 'two', 'four', 'six', 'eight', 'ten']

plt.xticks(ticks, labels)   # assign labels to ticks
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Create a figure
fig = plt.figure()

# Add a subplot (1 row, 1 column, 1st plot)
ax = fig.add_subplot(1, 1, 1)

# Generate random cumulative sum data
data = np.random.randn(1000).cumsum()

# Plot the data
ax.plot(data)

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create figure and axes
fig, ax = plt.subplots(figsize=(6,4))

# Plot
ax.plot(x, y)

# 1. Set x-axis range
ax.set_xlim(0, 10)

# 2. Set tick positions
ticks = [0, 2, 4, 6, 8, 10]
ax.set_xticks(ticks)

# 3. Set tick labels with rotation & fontsize
ax.set_xticklabels(['zero', 'two', 'four', 'six', 'eight', 'ten'],
                   rotation=45, fontsize=12)

# Title
ax.set_title("Example: Custom xticks & labels", fontsize=14)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
x = np.linspace(0, 2*np.pi, 200)   # 0 to 2π
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)

# Create figure and axes
fig, ax = plt.subplots(figsize=(8,5))

# Plot multiple waves with labels
ax.plot(x, y1, label="sin(x)")
ax.plot(x, y2, label="cos(x)")
ax.plot(x, y3, label="tan(x)")

# Limit y-range to avoid extreme tan values
ax.set_ylim(-5, 5)

# Set x-axis ticks and labels
ticks = [0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi]
labels = ['0', 'π/2', 'π', '3π/2', '2π']
ax.set_xticks(ticks)
ax.set_xticklabels(labels, rotation=45, fontsize=12)

# Title
ax.set_title("Sine, Cosine, and Tangent Waves", fontsize=14)

# Add legend
ax.legend(loc="upper right", fontsize=10)

plt.show()
```

Saving Plots to a File

Filename: "trig_functions.png" → can be .png, .jpg, .pdf, .svg, etc.

dpi=300 → high resolution, good for printing.

bbox_inches='tight' → trims extra whitespace around the figure.

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
x = np.linspace(0, 2*np.pi, 200)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)

# Create figure and axes
fig, ax = plt.subplots(figsize=(8,5))

# Plot
ax.plot(x, y1, label="sin(x)")
ax.plot(x, y2, label="cos(x)")
ax.plot(x, y3, label="tan(x)")
ax.set_ylim(-5, 5)

# Set ticks
ticks = [0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi]
labels = ['0', 'π/2', 'π', '3π/2', '2π']
ax.set_xticks(ticks)
ax.set_xticklabels(labels, rotation=45, fontsize=12)

# Title + legend
ax.set_title("Sine, Cosine, and Tangent Waves", fontsize=14)
ax.legend(loc="upper right")

# ---- Save plot to file ----
plt.savefig("trig_functions.pdf", dpi=300, bbox_inches='tight')

# Optional: also show
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load dataset directly from UCI repository (Iris CSV file)
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
iris = pd.read_csv(url)

# Create subplots: 1 row, 2 columns
fig, ax = plt.subplots(1, 2, figsize=(12, 5), constrained_layout=True)
'''constrained_layout=True automatically adjusts the spacing between subplots so that:

Axis labels don’t overlap with neighboring plots

Legends and titles don’t get cut off

Everything fits neatly inside the figure
'''
# Left subplot: Scatter plot
for species, group in iris.groupby("species"):
    ax[0].scatter(group["sepal_length"], group["sepal_width"], label=species)

ax[0].set_title("Sepal Length vs Width", fontsize=12)
ax[0].set_xlabel("Sepal Length (cm)")
ax[0].set_ylabel("Sepal Width (cm)")
ax[0].legend(title="Species")
ax[0].set_xticks(np.arange(4, 9, 1))
ax[0].set_yticks(np.arange(2, 5, 0.5))
ax[0].set_xticklabels(np.arange(4, 9, 1), rotation=45, fontsize=10)

# Right subplot: Histogram
for species, group in iris.groupby("species"):
    ax[1].hist(group["petal_length"], bins=10, alpha=0.6, label=species)

ax[1].set_title("Distribution of Petal Length", fontsize=12)
ax[1].set_xlabel("Petal Length (cm)")
ax[1].set_ylabel("Frequency")
ax[1].legend(title="Species")
ax[1].set_xticks(np.arange(1, 8, 1))
ax[1].set_xticklabels(np.arange(1, 8, 1), rotation=30, fontsize=10)

# Save the figure
plt.savefig("iris_visualization.png", dpi=300, bbox_inches='tight')
plt.show()
```
